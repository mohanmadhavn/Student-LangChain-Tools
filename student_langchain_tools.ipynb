```python
from langchain_core.tools import tool
@tool
def calculator(expression : str) -> str :
  """Calculate a mathematical expression."""
  try:
    result=eval(expression)
    return str(result)
  except Exception as e :
    return f"Calculation error : {e}"
calculator.invoke("85+72")
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model=LLM,
    tools=[
        calculator
    ]
)
```

```python
result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is 85 + 72?"
        }
    ]
})

print(result["messages"][-1].content)
```

```python
from langchain_core.tools import tool

@tool
def get_student_info(student_id: str) -> str:
    """Get a student's name and department using their student ID."""

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute("""
        SELECT name, department
        FROM students
        WHERE student_id = ?
    """, (student_id,))

    row = cursor.fetchone()

    conn.close()

    if row is None:
        return f"No student found with ID {student_id}"

    name, department = row

    return f"Name: {name}\nDepartment: {department}"
```

```python
@tool
def get_student_marks(student_id: str) -> str:
    """Get a student's Python, Database, AI, and Web marks using their student ID."""

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute("""
        SELECT python, database, ai, web
        FROM students
        WHERE student_id = ?
    """, (student_id,))

    row = cursor.fetchone()

    conn.close()

    if row is None:
        return f"No student found with ID {student_id}"

    python_mark, database_mark, ai_mark, web_mark = row

    return (
        f"Python: {python_mark}\n"
        f"Database: {database_mark}\n"
        f"AI: {ai_mark}\n"
        f"Web: {web_mark}"
    )
```

```python
print(get_student_marks.invoke("22CS045"))
```

```python
@tool
def get_passing_rules() -> str:
    """Get the university rules required for a student to pass."""

    return (
        "University passing rules:\n"
        "1. Minimum overall average: 40%\n"
        "2. Minimum mark in each subject: 35"
    )
```

```python
print(get_passing_rules.invoke(""))
```

```python
tools = [
    get_student_info,
    get_student_marks,
    calculator,
    get_passing_rules
]
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=LLM,
    tools=tools
)
```

```python
question = "What is the name and department of student 22CS045?"

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
})

print(result["messages"][-1].content)
```

```python
system_prompt = """
You are a university student information assistant.

You have four tools:

1. get_student_info
   Use this to obtain a student's name and department.

2. get_student_marks
   Use this to obtain Python, Database, AI, and Web marks.

3. calculator
   Use this for arithmetic such as totals, averages, and percentages.

4. get_passing_rules
   Use this to obtain the university's passing requirements.

Rules:
- Never guess student information or marks.
- Use the appropriate tools to obtain required information.
- If total or average marks are requested, use the calculator.
- If passing eligibility is requested, obtain both the student's marks
  and the university passing rules.
- Check both passing conditions:
    * overall average must be at least 40
    * every subject mark must be at least 35
- You decide which tools are needed. Do not assume a fixed workflow.
- Give the user a concise final answer.
"""

agent = create_agent(
    model=LLM,
    tools=tools,
    system_prompt=system_prompt
)
```

```python
question = """
I am 22CS045. Tell me my name, department, total marks,
average marks, and whether I satisfy the university
passing requirements.
"""

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
})

print(result["messages"][-1].content)
```

```python
def ask_student_agent(question):
    result = agent.invoke({
        "messages": [
            {
                "role": "user",
                "content": question
            }
        ]
    })

    return result["messages"][-1].content
```

```python
print(ask_student_agent(
    "What is the name and department of 22CS045?"
))
```
